```python
# print the average shortlisting rate of group A and group B candidates for each AI model version
for model_version in merged_df["AI_Model_Version"].unique():
    group_a_candidates = merged_df[(merged_df["AI_Model_Version"] == model_version) & (merged_df["Demographic_Group"] == "Group A")]
    group_b_candidates = merged_df[(merged_df["AI_Model_Version"] == model_version) & (merged_df["Demographic_Group"] == "Group B")]
    
    # print(group_a_candidates)
    # print(group_b_candidates)
    group_a_shortlisting_rate = group_a_candidates[group_a_candidates["Shortlisted"] == "Yes"].shape[0] / group_a_candidates.shape[0]
    group_b_shortlisting_rate = group_b_candidates[group_b_candidates["Shortlisted"] == "Yes"].shape[0] / group_b_candidates.shape[0]

    print(f"AI Model Version: {model_version}")
    print(f"Total Group A Candidates: {group_a_candidates.shape[0]}, Total Group B Candidates: {group_b_candidates.shape[0]}, Total Candidates: {group_a_candidates.shape[0] + group_b_candidates.shape[0]}")
    print(f"Total Shortlisted Group A Candidates: {group_a_candidates[group_a_candidates['Shortlisted'] == 'Yes'].shape[0]}, Total Shortlisted Group B Candidates: {group_b_candidates[group_b_candidates['Shortlisted'] == 'Yes'].shape[0]}")
    print(f"Group A Shortlisting Rate: {group_a_shortlisting_rate:.2f}")
    print(f"Group B Shortlisting Rate: {group_b_shortlisting_rate:.2f}")
    print()
```

```python
# plot pie chart for the first count of group A and group B candidates in total and for each AI model version
import matplotlib.pyplot as plt

for model_version in merged_df["AI_Model_Version"].unique():
    group_a_candidates = merged_df[(merged_df["AI_Model_Version"] == model_version) & (merged_df["Demographic_Group"] == "Group A")]
    group_b_candidates = merged_df[(merged_df["AI_Model_Version"] == model_version) & (merged_df["Demographic_Group"] == "Group B")]
    
    labels = ["Group A", "Group B"]
    sizes = [group_a_candidates.shape[0], group_b_candidates.shape[0]]
    
    plt.figure(figsize=(6, 6))
    plt.pie(sizes, labels=labels, autopct="%1.1f%%", startangle=90)
    plt.title(f"Candidate Distribution for AI Model Version: {model_version}")
    plt.axis("equal")
    plt.show()
```

```python
from scipy.stats import chi2_contingency

table = pd.crosstab(
    merged_df['Demographic_Group'],
    merged_df['College_Tier']
)

chi2, p, dof, expected = chi2_contingency(table)

print("Chi-square statistic:", chi2)
print("p-value:", p)
print("Degrees of freedom:", dof)

if p < 0.05:
    print("There is a statistically significant association between Demographic Group and College Tier.")
else:
    print("There is no statistically significant association between Demographic Group and College Tier.")
```
